### 2. Magyarországi korfája
Készíts egy korfához megfelelő adaszetet a "Kor", "Nem" és egy általad választott év oszlop használatával.

```python
df = pd.read_csv('https://github.com/everybitmihaly/elte_dh_adatviz_minor/raw/refs/heads/main/data/nepesseg_korev_es_nem_szerint.csv')
df.head(5)
```

```python
df = df[['Korév', 'Nem', '2020']].copy()  # Válaszd ki a szükséges oszlopokat, pl 'Korév', 'Nem', '2020'
```

```python
# Az apply funkcióval tudunk oszlopokra függvényt alkalmazni
# Az alábbi függvény egy életkort bemeneti paraméterként vár, és
# visszaadja a hozzá tartozó 5 éves életkori sávot

def create_age_range(age):

    if isinstance(age, str):
        age = age.replace('+', '')
        age = int(age)

    range_min = (age - (age % 5))
    range_max = range_min + 4
    return f"{range_min}-{range_max}"

df['range'] = df['Korév'].apply(create_age_range)
df
```

```python
# A coportosítás után a pivot() módszerrel alakítjuk át a táblát
# úgy, hogy a sorok az életkori sávok legyenek,
# az oszlopok a nemek, az értékek pedig a 2020-as népességszámok.

final = df[['range', 'Nem', '2020']] \
    .groupby(['range', "Nem"]).sum() \
    .reset_index() \
    .pivot(index='range', columns='Nem', values='2020')

final
```

```python
# Végül a férfiak értékeit negatívvá tesszük, hogy a korfa formátumba beilleszthető legyen
final['férfi'] = final['férfi'].apply(lambda x: x * -1)
final
```

```python
final.to_csv('')
```
